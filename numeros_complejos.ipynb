```python
import matplotlib.pyplot as plt
import cmath # Librería para funciones matemáticas complejas avanzadas

print("Seleccione una operación:")
print("1. Sumar")
print("2. Restar")
print("3. Multiplicar")
print("4. Dividir")
print("5. Potenciación")
print("6. Radicación")
print("7. Módulo")
print("8. Argumento angular")
print("9. Gráfica en plano Argand")

# En Python, los complejos se ingresan como: 2+3j
print("\nNota: Ingrese los números complejos usando 'j'. Ejemplo: 2+3j o 5j")

try:
    opcion = int(input('Ingrese su opcion: '))
    
    if opcion in [1, 2, 3, 4]:
        z1 = complex(input('Ingrese el primer numero complejo: '))
        z2 = complex(input('Ingrese el segundo numero complejo: '))
        
        if opcion == 1:
            print(f"Resultado de la suma: {z1+z2}")
        elif opcion == 2:
            print(f"Resultado de la resta: {z1-z2}")
        elif opcion == 3:
            print(f"Resultado de la multiplicacion: {z1*z2}")
        elif opcion == 4:
            if z2 != 0:
                print(f"Resultado de la division: {z1/z2}")
            else:
                print("Error: Division por cero.")
                
    elif opcion == 5:
        z1 = complex(input('Ingrese el numero complejo: '))
        exponente = float(input('Ingrese el exponente: '))
        print(f"Resultado de la potenciacion: {z1**exponente}")
        
    elif opcion == 6:
        z1 = complex(input('Ingrese el numero complejo: '))
        indice = float(input('Ingrese el indice del radical: '))
        if indice != 0:
            resultado = z1 ** (1/indice)
            print(f"Resultado de la radicacion: {resultado}")
        else:
            print("Error: El indice no puede ser cero.")
            
    elif opcion == 7:
        z1 = complex(input('Ingrese el numero complejo: '))
        print(f"Resultado del modulo: {abs(z1)}")
        
    elif opcion == 8:
        z1 = complex(input('Ingrese el numero complejo: '))
        # cmath.phase devuelve el argumento en radianes
        print(f"Resultado del argumento angular: {cmath.phase(z1)}")
        
    elif opcion == 9:
        z1 = complex(input('Ingrese el numero complejo: '))
        plt.figure()
        plt.plot(z1.real, z1.imag, 'ro') # 'ro' es círculo rojo
        plt.axhline(0, color='black', linewidth=0.5) # Eje X
        plt.axvline(0, color='black', linewidth=0.5) # Eje Y
        plt.xlabel('Parte Real')
        plt.ylabel('Parte Imaginaria')
        plt.title('Gráfica en el plano Argand')
        plt.grid(True)
        plt.show()
        
    else:
        print("Opción no válida.")

except ValueError:
    print("Error: Entrada no válida. Asegúrese de usar el formato correcto (ej: 1+2j).")
```
